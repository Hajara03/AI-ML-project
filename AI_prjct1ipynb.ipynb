#Handling Outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
num_col=df.select_dtypes(include='number').columns
num_col
```

```python
plt.figure(figsize=(15,14))
for i,col in enumerate(num_col):
  plt.subplot((len(num_col)+1)//2,2,i+1)
  sns.boxplot(x=df[col])
  plt.title(f'Boxplot of {col}')
plt.tight_layout()
plt.show()
```

```python
outlier_col=['No of Bedrooms', 'No of Bathrooms','Flat Area (in Sqft)','Lot Area (in Sqft)', 'Area of the House from Basement (in Sqft)','Basement Area (in Sqft)','Living Area after Renovation (in Sqft)','Lot Area after Renovation (in Sqft)']
def handle_outlier(outlier_col):
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1

  lb=Q1-1.5*IQR
  ub=Q3+1.5*IQR
  df[col]=df[col].clip(upper=ub,lower=lb)

for col in outlier_col:
  handle_outlier(col)
```
